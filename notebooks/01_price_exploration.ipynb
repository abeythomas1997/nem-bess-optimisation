```python
import csv
import pandas as pd

rows = []

with open(RAW_FILE, "r", newline="") as f:
    reader = csv.reader(f)

    for row in reader:
        if row[0] == "I" and row[1] == "DISPATCH" and row[2] == "PRICE":
            columns = row[4:]

        elif row[0] == "D" and row[1] == "DISPATCH" and row[2] == "PRICE":
            rows.append(row[4:])
```

```python
df = pd.DataFrame(rows, columns=columns)

df.head()
```

```python
df = df[["SETTLEMENTDATE", "REGIONID", "RRP"]]

df.head()
```

```python
df_vic = df[df["REGIONID"] == "VIC1"].copy()

df_vic.head()
```

```python
df_vic["SETTLEMENTDATE"] = pd.to_datetime(df_vic["SETTLEMENTDATE"])
df_vic["RRP"] = pd.to_numeric(df_vic["RRP"])

df_vic.info()
```
